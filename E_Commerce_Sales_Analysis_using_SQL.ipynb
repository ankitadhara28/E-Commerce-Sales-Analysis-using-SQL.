```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('Superstore.csv', encoding='latin1')
df.head()
```

```python
import sqlite3
# create connection
conn = sqlite3.connect(':memory:')
# convert dataframe to SQL table
df.to_sql('sales', conn, index=False, if_exists='replace')
```

```python
pd.read_sql("SELECT * FROM sales LIMIT 5;", conn)
```

```python
pd.read_sql("SELECT * FROM sales;", conn)
```

```python
# Calculate total sales (overall revenue of the company):
pd.read_sql("SELECT SUM(Sales) AS Total_Sales FROM sales;", conn)
```

```python
# Find top 5 products by total sales:
# This query identifies which products generate the highest revenue
top_products = pd.read_sql("""
SELECT "Product Name", SUM(Sales) AS Total_Sales
FROM sales
GROUP BY "Product Name"
ORDER BY Total_Sales DESC
LIMIT 5;
""", conn)

top_products
```

```python
# Visualize top 5 products using a horizontal bar chart
# This makes it easier to compare product names and their sales values

plt.figure()
colors = ['#FF6F61', '#6B5B95', '#88B04B', '#F7CAC9', '#92A8D1']

# Create horizontal bar chart
plt.barh(top_products['Product Name'], top_products['Total_Sales'], color=colors)

# Add title and labels
plt.title("Top 5 Products by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Product Name")

# Invert y-axis so highest value appears on top
plt.gca().invert_yaxis()

# Display chart
plt.show()
```

```python
# Calculate total sales for each region:
# This helps identify which region contributes the most to overall revenue
region_data = pd.read_sql("""
SELECT Region, SUM(Sales) AS Total_Sales
FROM sales
GROUP BY Region
ORDER BY Total_Sales DESC;
""", conn)

region_data
```

```python
# Visualize total sales across different regions
# This chart helps identify which region contributes the most to overall revenue

plt.figure()
colors = ['#4CAF50', '#2196F3', '#FF9800', '#9C27B0']  # green, blue, orange, purple

# Create bar chart
plt.bar(region_data['Region'], region_data['Total_Sales'], color=colors)

# Add title and labels for clarity
plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")

# Display the chart
plt.show()
```

```python
# Calculate total profit for each product category:
# This helps identify which category is most profitable for the business
category_data = pd.read_sql("""
SELECT Category, SUM(Profit) AS Total_Profit
FROM sales
GROUP BY Category
ORDER BY Total_Profit DESC;
""", conn)

category_data
```

```python
# Visualize profit by category using a bar chart
# This chart helps compare profitability across different product categories

plt.figure()
colors = ['#4CAF50', '#2196F3', '#FF9800']  # green, blue, orange
plt.bar(category_data['Category'], category_data['Total_Profit'], color=colors)

plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")

plt.show()
```

```python
# Find products where total profit is negative (loss-making products)
# This helps identify which products are hurting the business and need attention

pd.read_sql("""
SELECT "Product Name", SUM(Profit) AS Total_Profit
FROM sales
GROUP BY "Product Name"
HAVING SUM(Profit) < 0
ORDER BY Total_Profit ASC;
""", conn)
```

```python
# Extract top 10 loss-making products
# This helps identify products causing the highest losses to the business

loss_products = pd.read_sql("""
SELECT "Product Name", SUM(Profit) AS Total_Profit
FROM sales
GROUP BY "Product Name"
HAVING SUM(Profit) < 0
ORDER BY Total_Profit ASC
LIMIT 10;
""", conn)

loss_products
```

```python
# STEP: Visualize top loss-making products using horizontal bar chart
# This highlights which products are causing the most financial loss

plt.figure()
colors = ['#B71C1C', '#C62828', '#D32F2F', '#E53935', '#F44336',
          '#EF5350', '#E57373', '#EF9A9A', '#FFCDD2', '#FFEBEE']

# Create horizontal bar chart
plt.barh(loss_products['Product Name'], loss_products['Total_Profit'], color=colors)

# Add title and labels
plt.title("Top 10 Loss-Making Products")
plt.xlabel("Total Profit (Negative = Loss)")
plt.ylabel("Product Name")

# Show biggest loss at top
plt.gca().invert_yaxis()

# Display chart
plt.show()
```
